# Project 2: Image Recognition with MobileNetV2

This project performs image recognition using transfer learning with a pre-trained **MobileNetV2** model on ImageNet.

## 1. Import Libraries

```python
import os
import time
import urllib.request
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input, decode_predictions
from tensorflow.keras.utils import load_img, img_to_array

print("TensorFlow version:", tf.__version__)
```

## 2. Load Pre-trained MobileNetV2 Model

```python
image_model = MobileNetV2(weights="imagenet")
print("MobileNetV2 model loaded successfully.")
```

## 3. Load Image Input

You can use the sample image (Option A) or load your own custom image by commenting out Option A and uncommenting Option B.

### Option A: Sample / Fallback Image (Default)

```python
# OPTION A: Use default sample image (Default)
FALLBACK_IMAGE_URL = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/baboon.jpg"
FALLBACK_IMAGE_PATH = "sample_image.jpg"

image_path = FALLBACK_IMAGE_PATH if os.path.exists(FALLBACK_IMAGE_PATH) else None
if not image_path:
    urllib.request.urlretrieve(FALLBACK_IMAGE_URL, FALLBACK_IMAGE_PATH)
    image_path = FALLBACK_IMAGE_PATH

image = load_img(image_path, target_size=(224, 224))

plt.figure(figsize=(6, 6))
plt.imshow(image)
plt.axis("off")
plt.title(f"Input Sample Image ({image_path})")
plt.show()
```

### Option B: Custom User Image Input (Alternative)

## 4. Preprocess Image

```python
image_array = img_to_array(image)
image_batch = np.expand_dims(image_array, axis=0)
processed_image = preprocess_input(image_batch)

print("Original shape:", image_array.shape)
print("Batch shape:", image_batch.shape)
```

## 5. Model Prediction & Decoding

```python
start_time = time.time()
predictions = image_model.predict(processed_image, verbose=0)
elapsed_ms = (time.time() - start_time) * 1000

decoded_predictions = decode_predictions(predictions, top=5)[0]

print(f"Prediction took {elapsed_ms:.1f} ms")
print("\nTop Predictions:")
for _, label, probability in decoded_predictions:
    print(f"{label}: {probability * 100:.2f}%")
```

## 6. Conclusion

MobileNetV2 accurately classifies images using features learned from the ImageNet dataset.
